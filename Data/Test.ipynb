```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the CSV
train_df = pd.read_csv('sign_mnist_train.csv')

# Separate labels and pixels
labels = train_df['label'].values
pixels = train_df.drop('label', axis=1).values

# Reshape from (N, 784) → (N, 28, 28) for image format
images = pixels.reshape(-1, 28, 28).astype(np.uint8)

# Visualize one sample
plt.imshow(images[6], cmap='gray')
plt.title(f"Label: {labels[6]}")
plt.show()
```
